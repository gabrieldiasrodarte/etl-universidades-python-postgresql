```python
import pandas as pd #utilizada para ler, manipular, tratar e analisar dados.
import psycopg2 as ps2 #utilizada para fazer a conexão com banco de dados
```

```python
#leitura
caminho_arquivo = "sot.json"
df = pd.read_json(caminho_arquivo, encoding="utf-8")# leio o arquivo e insiro na variável df
df.head() # mostra as primeiras colunas do meu dataframe
```

```python
#nessa primeira parte eu realizo a conexão com o banco de dados PostgreSQL, passando os parâmetros 
conexao = ps2.connect( 
    dbname='universidades',
    user='postgres',
    password='', 
    host='localhost',
    port='5432')

cursor = conexao.cursor() # criamos a conexão python com o servidor PostgreSQL, e o cursor é responsável por enviar comandos sql ao banco  para depois executá-los
cursor.execute('delete from informacoes_universidade') # aqui eu deleto dados do banco para depois inserir mais, não é a maneira mais recomendada 

for indice, coluna_df in df.iterrows(): #percorro meu dataframe 
    #insiro as informações do meu dataframe na tabela informacoes_universidade nas seguintes colunas
    cursor.execute("""insert into informacoes_universidade( 
        nome_orgao,
        titulo,
        local,
        data_criacao
        
    )Values(%s,%s,%s,%s)""", #essa parte cada %s vai receber os valores de cada coluna percorrendo linha a linha
(       #essa parte passo os nomes das minhas colunas do dataframe que são os valores que prencheram o banco
         coluna_df["nome_orgao"],
         coluna_df["titulo"],
         coluna_df["Local"],
         coluna_df["data_criacao"])
    )
conexao.commit()# aqui eu valido a operação confirmando ela
# e fecho o cursor e a conexão isso auxilia em evitar problemas futuros
cursor.close()
conexao.close()



```
